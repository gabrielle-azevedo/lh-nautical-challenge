```python
import pandas as pd
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
```

```python
df_custos = pd.read_json("custos_importacao.json")
print("Tamanho:", df_custos.shape)
```

```python
print("Colunas:", df_custos.columns.tolist())
print("\nTipos de dados:")
print(df_custos.dtypes)
print("\nPrimeiras linhas:")
df_custos.head()
```

```python
resumo_id = pd.DataFrame({
    "Métrica": [
        "Total de registros",
        "IDs únicos",
        "IDs nulos",
        "IDs duplicados"
    ],
    "Valor": [
        len(df_custos),
        df_custos["product_id"].nunique(),
        df_custos["product_id"].isnull().sum(),
        df_custos["product_id"].duplicated().sum()
    ]
})
print("TABELA — PRODUCT_ID")
print(resumo_id.to_string(index=False))
```

```python
resumo_name = pd.DataFrame({
    "Métrica": [
        "Total de registros",
        "Nomes únicos",
        "Nomes nulos",
        "Nomes em branco"
    ],
    "Valor": [
        len(df_custos),
        df_custos["product_name"].nunique(),
        df_custos["product_name"].isnull().sum(),
        (df_custos["product_name"].str.strip() == "").sum()
    ]
})
print("TABELA — PRODUCT_NAME")
print(resumo_name.to_string(index=False))
```

```python
resumo_category = pd.DataFrame({
    "Métrica": [
        "Total de registros",
        "Categorias únicas",
        "Nulos",
        "Em branco"
    ],
    "Valor": [
        len(df_custos),
        df_custos["category"].nunique(),
        df_custos["category"].isnull().sum(),
        (df_custos["category"].str.strip() == "").sum()
    ]
})
print("TABELA — CATEGORY")
print(resumo_category.to_string(index=False))

print("\nCategorias encontradas:")
print(df_custos["category"].value_counts())
```

```python
print("TABELA — HISTORIC_DATA")
print("\nTipo da coluna:", df_custos["historic_data"].dtype)
print("\nPrimeiros valores:")
print(df_custos["historic_data"].head(10))
print("\nNulos:", df_custos["historic_data"].isnull().sum())
```

```python
print(type(df_custos["historic_data"][0]))
```

```python
import json

# Se estiver como string, converter para lista
df_custos["historic_data"] = df_custos["historic_data"].apply(
    lambda x: json.loads(x) if isinstance(x, str) else x
)

print("Conversão concluída!")
print("Exemplo do primeiro registro:")
print(df_custos["historic_data"][0])
```

```python
def pegar_custo_recente(historico):
    try:
        # Ordenar por data e pegar o mais recente
        ordenado = sorted(historico, key=lambda x: pd.to_datetime(x["start_date"], dayfirst=True))
        return ordenado[-1]["usd_price"]
    except:
        return None

df_custos["custo_atual_usd"] = df_custos["historic_data"].apply(pegar_custo_recente)

print("Custo mais recente extraído!")
print(df_custos[["product_id", "product_name", "custo_atual_usd"]].head())
```

```python
registros = []

for _, row in df_custos.iterrows():
    for entrada in row["historic_data"]:
        registros.append({
            "product_id": row["product_id"],
            "product_name": row["product_name"],
            "category": row["category"],
            "start_date": entrada["start_date"],
            "usd_price": entrada["usd_price"]
        })

df_custos_expandido = pd.DataFrame(registros)
df_custos_expandido["start_date"] = pd.to_datetime(
    df_custos_expandido["start_date"], dayfirst=True
)

print("Histórico expandido!")
print("Total de registros:", len(df_custos_expandido))
df_custos_expandido.head()
```

```python
resumo_custos = pd.DataFrame({
    "Métrica": [
        "Total de registros",
        "Produtos únicos",
        "Data mais antiga",
        "Data mais recente",
        "Custo mínimo (USD)",
        "Custo máximo (USD)",
        "Custo médio (USD)"
    ],
    "Valor": [
        len(df_custos_expandido),
        df_custos_expandido["product_id"].nunique(),
        df_custos_expandido["start_date"].min().strftime("%d/%m/%Y"),
        df_custos_expandido["start_date"].max().strftime("%d/%m/%Y"),
        f"$ {df_custos_expandido['usd_price'].min():,.2f}",
        f"$ {df_custos_expandido['usd_price'].max():,.2f}",
        f"$ {df_custos_expandido['usd_price'].mean():,.2f}"
    ]
})

print("TABELA — CUSTOS DE IMPORTAÇÃO")
print(resumo_custos.to_string(index=False))
```

```python
# Versão resumida com custo atual
df_custos[["product_id", "product_name", "category", "custo_atual_usd"]].to_csv(
    "custos_resumido.csv", index=False
)

# Versão expandida com todo o histórico
df_custos_expandido.to_csv("custos_expandido.csv", index=False)

print("Bases de custos salvas!")
```
