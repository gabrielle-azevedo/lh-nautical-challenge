```python
import pandas as pd
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
df = pd.read_csv("vendas_2023_2024.csv")
```

```python
print("Tamanho:", df.shape)
print("\nColunas:", df.columns.tolist())
print("\nTipos de dados:")
print(df.dtypes)
print("\nPrimeiras linhas:")
df.head()
```

```python
print("Valores nulos:")
print(df.isnull().sum())

print("\nDuplicatas:", df.duplicated().sum())
```

```python
df.describe()
```

```python
# Período coberto pelas vendas
print("Data mais antiga:", df["sale_date"].min())
print("Data mais recente:", df["sale_date"].max())

# Quantidade vendida
print("\nEstatísticas de quantidade:")
print(df["qtd"].describe())

# Quantidade total vendida
print("\nEstatísticas de total:")
print(df["total"].describe())

# Quantos clientes únicos
print("\nClientes únicos:", df["id_client"].nunique())

# Quantos produtos únicos vendidos
print("\nProdutos únicos:", df["id_product"].nunique())
```

```python
resumo_qtd = pd.DataFrame({
    "Métrica": [
        "Total de registros",
        "Quantidade mínima",
        "Quantidade máxima",
        "Média",
        "Zeros",
        "Negativos"
    ],
    "Valor": [
        len(df),
        df["qtd"].min(),
        df["qtd"].max(),
        round(df["qtd"].mean(), 2),
        (df["qtd"] == 0).sum(),
        (df["qtd"] < 0).sum()
    ]
})

print("TABELA — QUANTIDADE (qtd)")
print(resumo_qtd.to_string(index=False))
```

```python
resumo_total = pd.DataFrame({
    "Métrica": [
        "Total de registros",
        "Valor mínimo",
        "Valor máximo",
        "Média",
        "Soma total (faturamento)",
        "Zeros",
        "Negativos"
    ],
    "Valor": [
        len(df),
        df["total"].min(),
        df["total"].max(),
        round(df["total"].mean(), 2),
        round(df["total"].sum(), 2),
        (df["total"] == 0).sum(),
        (df["total"] < 0).sum()
    ]
})

print("TABELA — VALOR TOTAL (total)")
print(resumo_total.to_string(index=False))
```

```python
ranking = df.groupby("id_product").agg(
    total_qtd=("qtd", "sum"),
    total_faturado=("total", "sum")
).reset_index().sort_values("total_faturado", ascending=False)

print("\n Top 10 produtos — quantidade e valor:")
print(ranking.head(10).to_string(index=False))
```

```python
resumo_clientes = pd.DataFrame({
    "Métrica": [
        "Total de registros",
        "Clientes únicos",
        "Clientes nulos",
        "Média de compras por cliente"
    ],
    "Valor": [
        len(df),
        df["id_client"].nunique(),
        df["id_client"].isnull().sum(),
        round(df.groupby("id_client").size().mean(), )
    ]
})

print("TABELA — CLIENTES")
print(resumo_clientes.to_string(index=False))
```

```python
resumo_produtos = pd.DataFrame({
    "Métrica": [
        "Total de registros",
        "Produtos únicos vendidos",
        "Produtos nulos",
        "Produto mais vendido (id)"
    ],
    "Valor": [
        len(df),
        df["id_product"].nunique(),
        df["id_product"].isnull().sum(),
        df["id_product"].value_counts().idxmax()
    ]
})

print("TABELA — PRODUTOS")
print(resumo_produtos.to_string(index=False))
```

```python
df["sale_date"] = pd.to_datetime(df["sale_date"], format='mixed', dayfirst=True)

resumo_datas = pd.DataFrame({
    "Métrica": [
        "Total de registros",
        "Data mais antiga",
        "Data mais recente",
        "Período coberto (dias)",
        "Datas nulas"
    ],
    "Valor": [
        len(df),
        df["sale_date"].min().strftime("%d/%m/%Y"),
        df["sale_date"].max().strftime("%d/%m/%Y"),
        (df["sale_date"].max() - df["sale_date"].min()).days,
        df["sale_date"].isnull().sum()
    ]
})

print("TABELA — DATAS")
print(resumo_datas.to_string(index=False))
```

```python
print("Duplicatas antes:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicatas depois:", df.duplicated().sum())
```

```python
print("Nulos por coluna:")
print(df.isnull().sum())

# Remover linhas sem id, id_client ou id_product
df = df.dropna(subset=["id", "id_client", "id_product"])

# Remover linhas sem data
df = df.dropna(subset=["sale_date"])
```

```python
# Garantir que a data está no formato correto
df["sale_date"] = pd.to_datetime(df["sale_date"])

# Garantir que qtd e total são números
df["qtd"] = pd.to_numeric(df["qtd"], errors="coerce")
df["total"] = pd.to_numeric(df["total"], errors="coerce")

# Garantir que ids são inteiros
df["id"] = df["id"].astype(int)
df["id_client"] = df["id_client"].astype(int)
df["id_product"] = df["id_product"].astype(int)
```

```python
print("Data mais antiga:", df["sale_date"].min().strftime("%d/%m/%Y"))
print("Data mais recente:", df["sale_date"].max().strftime("%d/%m/%Y"))

# Remover datas fora do período esperado (2023-2024)
df = df[df["sale_date"].dt.year.isin([2023, 2024])]
print("Registros após filtro de período:", len(df))
```
